# Scientific Python and PyROOT demonstration

Select the **Python 3 (ipykernel)** kernel and run the cells in order. This example plots a sine wave with NumPy/Matplotlib and creates a Gaussian histogram with PyROOT. All three packages are included in the notebook image.

```python
import numpy as np
import matplotlib.pyplot as plt
import ROOT
```

```python
x = np.linspace(0, 2*np.pi, 1000)
y = np.sin(x)
```

```python
plt.plot(x, y)
```

```python
ROOT.EnableImplicitMT()
```

```python
# Crea un istogramma
h = ROOT.TH1F("h", "Test ROOT;X;Entries", 100, -5, 5)

# Riempi con numeri casuali gaussiani
h.FillRandom("gaus", 10000)

# Visualizza nel notebook
canvas = ROOT.TCanvas("canvas", "ROOT Test", 800, 600)
```

```python
h.Draw()
```

```python
canvas.Draw()
```
