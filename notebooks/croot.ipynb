# ROOT C++ demonstration

Select the **ROOT C++** kernel and run the cells in order. This example generates a Gaussian histogram and renders it with ROOT. The notebook image installs ROOT; confirm the C++ kernel is listed with `jupyter kernelspec list` before running.

```c++
#include <iostream>

std::cout << "Ciao da ROOT C++!" << std::endl;
```

```c++
#include <TH1F.h>
#include <TCanvas.h>
#include <TRandom.h>

TH1F *h = new TH1F(
    "h",
    "Distribuzione gaussiana;X;Conteggi",
    100,
    -5,
    5
);

for (int i = 0; i < 10000; i++) {
    h->Fill(gRandom->Gaus(0, 1));
}

TCanvas *c = new TCanvas("c", "Test ROOT", 800, 600);

h->Draw();
c->Draw();
```
